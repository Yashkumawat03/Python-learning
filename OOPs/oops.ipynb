```python
### Instant variables an Methods
class Dog:
    ## Constructor
    def __init__(self,name,age):
        self.name = name 
        self.age = age

## Create objects
dog1 = Dog("Yash",22)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2= Dog("Yuvraj",22)
print(dog2.name)
print(dog2.age)
```

```python
## Modelling a bank account

## Defining a class for bank account 
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self,amount):
        self.balance = self.balance + amount
        print(f"{amount} is deposited. New balance is {self.balance}.")

    def withdrawl(self,amount):
        if amount > self.balance:
            print(f"Insufficience Balance")
        
        else:
            self.balance = self.balance - amount
            print(f"{amount} is withdrawl. New balance is {self.balance}.")


## Create an account
account = BankAccount("Yash",50000)
print(account.balance)
```

```python
## Call instance methods

account.deposit(100)
```

```python
account.withdrawl(500)
```
